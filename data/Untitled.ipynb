```python
from pathlib import Path
from huggingface_hub import hf_hub_download

DATA_DIR = Path("../data/raw")
DATA_DIR.mkdir(parents=True, exist_ok=True)

REPO_ID = "ethanchrstian/siaga-pantura-hazard-data"

files = [
    "flood_dataset.parquet",
    "drought_dataset.parquet",
    "risk_history.parquet",
    "districts.geojson",
    "district_centroids.csv",
    "population.csv",
]

for filename in files:
    path = hf_hub_download(
        repo_id=REPO_ID,
        filename=filename,
        repo_type="dataset",
        local_dir=str(DATA_DIR)
    )
    print("Berhasil:", filename)
    print("Lokasi:", path)
```

```python
import polars as pl

file_path = DATA_DIR / "flood_dataset.parquet"

df = pl.scan_parquet(file_path)

jumlah_baris = df.select(pl.len()).collect().item()

print("Jumlah baris:", f"{jumlah_baris:,}")
print("Memenuhi syarat 1 juta:", jumlah_baris >= 1_000_000)
```
